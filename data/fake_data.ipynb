```python
import pandas as pd
import numpy as np

fake_df = pd.read_csv("no_beta_engineered_data.csv")

# 15% sample from each class
sample_df = (
    fake_df.groupby("no_show", group_keys=False)
      .sample(frac=0.15, random_state=42)
)

# Create synthetic data
synthetic_parts = []

for cls, group in sample_df.groupby("no_show"):
    synthetic = pd.DataFrame(index=range(len(group)))
    synthetic["no_show"] = cls

    for col in group.columns:
        if col == "no_show":
            continue

        synthetic[col] = np.random.choice(
            group[col],
            size=len(group),
            replace=True
        )

    synthetic_parts.append(synthetic)

synthetic_df = pd.concat(synthetic_parts, ignore_index=True)

synthetic_df.to_csv("fake_choc_data.csv", index=False)
```
