```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
raw_sdoh = pd.read_csv('data/Non-Medical_Factor_Measures_for_Census_Tract__ACS_2017-2021.csv')
```

```python
raw_sdoh.head()
```

```python
cal_sdoh = raw_sdoh.where(raw_sdoh['StateAbbr'] == 'CA').dropna()
```

```python
oc_sdoh = cal_sdoh.where(cal_sdoh['CountyName'] == 'Orange County').dropna()
```

```python
oc_sdoh
```

```python
oc_proj = oc_sdoh[['Data_Value', 'MOE', 'TotalPopulation', 'LocationID', 'Short_Question_Text', 'Geolocation']]
```

```python
oc_proj
```

```python
oc_proj
```

```python
oc_values = oc_proj.pivot(index=['Geolocation', 'TotalPopulation'],columns='Short_Question_Text', values='Data_Value')
oc_values = oc_values.reset_index()
```

```python
oc_values['Scaled_Population'] = 0.02*(oc_values['TotalPopulation'])
oc_values
```

```python
oc_values['TotalPopulation'].max(), oc_values['TotalPopulation'].min(), oc_values['TotalPopulation'].median()
```

```python
import geopandas as gpd
from shapely import wkt
import pygris
# This line turns the string into a real map coordinate
df = gpd.GeoDataFrame(oc_values)
df['Geolocation'] = df['Geolocation'].apply(wkt.loads)
df.set_geometry('Geolocation', crs="EPSG:4326", inplace=True)
```

```python
import matplotlib.lines as mlines
import numpy as np
import contextily as ctx

def plot_on_oc (df, column):
    # ... [Keep your data prep logic] ...
    ca_counties = pygris.counties(state="CA", cb=True)
    oc_boundary = ca_counties[ca_counties.NAME == "Orange"].to_crs(epsg=3857)
    df = df.to_crs(epsg=3857)

    fig, ax = plt.subplots(figsize=(10, 8))

    # 1. Background Fill
    oc_boundary.plot(ax=ax, color='lightgray', edgecolor='black', linewidth=2, zorder=1)
    ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron, zorder=2)

    # 2. Plot Points
    df.plot(
        ax=ax, 
        column=column, 
        markersize='Scaled_Population',
        legend=True,
        # 'fontsize' is not valid here, so we remove it to avoid errors
        legend_kwds={'shrink': 0.6}, 
        alpha=0.7, 
        edgecolor='k',
        zorder=3
    )

    # 3. Manual Color Axis Font Scaling
    # fig.get_axes()[1] specifically targets the colorbar axis
    cax = fig.get_axes()[1]
    cax.tick_params(labelsize=16) # Size of the numbers
    cax.set_ylabel(f"SDOH Index: {column}", fontsize=22, labelpad=20) # Size of the label

    # 4. Size Legend (Axis)
    pop_min, pop_med, pop_max = df['Scaled_Population'].min(), df['Scaled_Population'].median(), df['Scaled_Population'].max()
    
    l1 = mlines.Line2D([], [], color='white', marker='o', markerfacecolor='gray',
                       markeredgecolor='k', markersize=np.sqrt(pop_min), label='1000', ls='')
    l2 = mlines.Line2D([], [], color='white', marker='o', markerfacecolor='gray',
                       markeredgecolor='k', markersize=np.sqrt(pop_med), label='5000', ls='')
    l3 = mlines.Line2D([], [], color='white', marker='o', markerfacecolor='gray',
                       markeredgecolor='k', markersize=np.sqrt(pop_max), label='15000', ls='')

    size_legend = ax.legend(handles=[l1, l2, l3], title="Population Size", 
                            loc='lower left', fontsize=14, title_fontsize=16,
                            frameon=True, borderpad=1.5, labelspacing=1.5)
    ax.add_artist(size_legend)

    # 5. Centered Figure Title
    fig.suptitle(f"{column} rates split by FIP codes \n across Orange County", fontsize=28, fontweight='bold', y=0.95)
    
    ax.set_axis_off()
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```

```python
columns = [
    'Aged 65 years or older',
 'Crowding',
 'Housing cost burden',
 'No broadband',
 'No high school diploma',
 'Poverty',
 'Racial or ethnic minority status',
 'Single-parent households',
 'Unemployment',
 ]
```

```python
for column in columns:
    plot_on_oc(df,column)
```
