```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
raw_sdoh = pd.read_csv('Non-Medical_Factor_Measures_for_Census_Tract__ACS_2017-2021.csv')
```

```python
raw_sdoh.head()
```

```python
cal_sdoh = raw_sdoh.where(raw_sdoh['StateAbbr'] == 'CA').dropna()
```

```python
oc_sdoh = cal_sdoh.where(cal_sdoh['CountyName'] == 'Orange County').dropna()
```

```python
oc_sdoh
```

```python
oc_proj = oc_sdoh[['Data_Value', 'MOE', 'TotalPopulation', 'LocationID', 'Short_Question_Text', 'Geolocation']]
```

```python
oc_proj
```

```python
oc_proj
```

```python
oc_values = oc_proj.pivot(index=['Geolocation', 'TotalPopulation'],columns='Short_Question_Text', values='Data_Value')
oc_values = oc_values.reset_index()
```

```python
oc_values['Scaled_Population'] = 0.02*(oc_values['TotalPopulation'])
oc_values
```

```python
import geopandas as gpd
from shapely import wkt
# This line turns the string into a real map coordinate
df = gpd.GeoDataFrame(oc_values)
df['Geolocation'] = df['Geolocation'].apply(wkt.loads)
df.set_geometry('Geolocation', crs="EPSG:4326", inplace=True)
```

```python
import pygris
import contextily as ctx

def plot_on_oc (df, column):
    # 1. Fetch all counties in CA, then filter for Orange
    # The 'cb=True' makes it a lighter file for faster plotting
    ca_counties = pygris.counties(state="CA", cb=True)

    # 2. Filter specifically for Orange County
    # Orange County's FIPS code is "059"
    oc_boundary = ca_counties[ca_counties.NAME == "Orange"]

    # Re-project everything to Web Mercator (EPSG:3857) to match basemaps
    oc_boundary = oc_boundary.to_crs(epsg=3857)
    df = df.to_crs(epsg=3857)

    # Updated plotting section:
    fig, ax = plt.subplots(figsize=(12, 9))

    # 1. Fill the county shape with a light color
    oc_boundary.plot(
        ax=ax, 
        color='lightgray',   # Change from None to a color to "fill" the shape
        edgecolor='black', 
        linewidth=2, 
        zorder=1             # Ensure it is at the bottom layer
    )

    # 2. Add basemap (ensure this is layered correctly)
    ctx.add_basemap(ax, source=ctx.providers.CartoDB.Positron, zorder=2)

    # 3. Plot your SDOH points on top
    df.plot(
        ax=ax, 
        column=column, 
        markersize='Scaled_Population',
        legend=True,
        legend_kwds={'label': f"SDOH index for {column}"},
        alpha=0.7, 
        edgecolor='k',
        zorder=3             # Ensure points are on top
    )

    plt.title(f"Orange County SDOH Index for {column}")
    ax.set_axis_off()
    plt.show()
```

```python
columns = [
    # 'Aged 65 years or older',
#  'Crowding',
#  'Housing cost burden',
#  'No broadband',
#  'No high school diploma',
#  'Poverty',
#  'Racial or ethnic minority status',
 'Single-parent households',
#  'Unemployment',
 ]
```

```python
for column in columns:
    plot_on_oc(df,column)
```
